# Logistic Regression

Predict disease from symptoms using the final dataset.

```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
SRC_DIR = PROJECT_ROOT / "src"
sys.path.insert(0, str(SRC_DIR))

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.pipeline import Pipeline
from nlp_medical_dataset.config.settings import FINAL_DATASET_CSV, REPORTS_DIR
from nlp_medical_dataset.models.ml_utils import load_medical_dataset, split_dataset, evaluate_model, save_metrics

df = load_medical_dataset(FINAL_DATASET_CSV)
df.head()
```

```python
x_train, x_test, y_train, y_test = split_dataset(df)
print(len(x_train), len(x_test))
```

```python
from sklearn.linear_model import LogisticRegression

model = Pipeline([
    ("tfidf", TfidfVectorizer(ngram_range=(1, 2), min_df=1)),
    ("clf", LogisticRegression(max_iter=1000))
])
```

```python
model.fit(x_train, y_train)
metrics = evaluate_model(model, x_test, y_test)
save_metrics(REPORTS_DIR / "logistic_regression_metrics.json", metrics)
metrics["accuracy"]
```
